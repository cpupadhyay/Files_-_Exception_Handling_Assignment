9. What is the difference between multithreading and multiprocessing?
   -> Multithreading and multiprocessing are both techniques used for concurrent execution of tasks, but they differ in how they handle tasks and system resources.
       ->Multithreading:
         ->In multithreading, multiple threads run within the same process. Threads share the same memory space, making it easier to share data.
         ->Suitable for I/O-bound tasks (such as reading from files, network operations) where the program spends a lot of time waiting for external resources.
         ->Due to Python's Global Interpreter Lock (GIL), only one thread can execute Python bytecode at a time, limiting the effectiveness of multithreading for CPU-bound tasks.
       ->Multiprocessing:
         ->In multiprocessing, each process runs independently with its own memory space. This approach creates separate processes that do not share memory, which allows better isolation.
         ->Suitable for CPU-bound tasks (such as computation-heavy tasks) because each process can run on a separate core, taking full advantage of multi-core processors.
         ->Unlike multithreading, multiprocessing does not face the GIL limitation.

12. What are the basic steps involved in exception handling in Python?
    -> In Python,exception handling is used to gracefully handle errors that occur during program execution without crashing the program.
       -> Try Block (`try`):
          -> Place the code that might cause an error inside the `try` block.
          -> If an error occurs, Python immediately stops executing the `try` block and jumps to the `except` block.
       -> Except Block (`except`):
          -> The `except` block handles the specific error or all errors.

       -> Else Block (Optional):
          -> If no exception occurs in the `try` block, the `else` block gets executed.
          -> It’s useful for running code only when the try block succeeds.
       -> Finally Block (Optional):
          -> This block always runs, whether an exception occurs or not.
          -> Commonly used for cleanup actions** like closing files, releasing resources, etc.

15. How does Python's garbage collection system work?
    -> Python’s garbage collection system is an automatic memory management mechanism that reclaims memory by cleaning up objects that are no longer needed — so you don’t have to do it manually.
       -> Reference Counting (Primary Mechanism):
          -> Every object in Python has a reference count — the number of variables referring to it.
          -> When an object’s reference count drops to zero, it means no part of the program is using it anymore.
          -> Python automatically deletes such objects to free memory.
      -> Cyclic Garbage Collector (Secondary Mechanism):
         -> Sometimes, two or more objects reference each other, forming a cycle.
         -> Even if nothing else refers to them, their reference count never drops to zero.
         -> Python’s garbage collector detects and cleans up these circular references.

18. What is the difference between os.fork() and multiprocessing in Python?
    -> Great question! Both `os.fork()` and the `multiprocessing` module in Python are used to create new processes, but they work differently and are suited for different use cases.
      -> `os.fork()` — Low-Level Process Creation
          -> What it does: Creates a child process by duplicating the current process.
          -> How it works: Directly uses the `fork()` system call from the OS (works only on Unix/Linux, not on Windows).
          -> Return value: Returns `0` in the child process, and the PID of the child in the parent process.
      -> `multiprocessing` Module — High-Level Interface
          -> What it does: Provides a cross-platform way to create and manage separate processes.
          -> How it works: Uses `fork()` on Unix, and `spawn()` on Windows (more compatible).
          -> Includes: Tools for inter-process communication, synchronization, shared memory, and pools of processes.

```python
#4. Write a Python script that reads from one file and writes its content to another file.
#Code
# File: copy_file.py

def copy_file(source_file, destination_file):
    try:
        with open(source_file, 'r') as src:
            content = src.read()

        with open(destination_file, 'w') as dest:
            dest.write(content)

        print(f"Contents copied from '{source_file}' to '{destination_file}' successfully.")

    except FileNotFoundError:
        print(f"The file '{source_file}' does not exist.")
    except Exception as e:
        print(f"An error occurred: {e}")

# Example usage
source = 'input.txt'         # Replace with your source file path
destination = 'output.txt'   # Replace with your destination file path

copy_file(source, destination)
```

```python
#5. How would you catch and handle division by zero error in Python?
#Code
def divide(a, b):
    try:
        result = a / b
        print(f"Result: {result}")
    except ZeroDivisionError:
        print("Error: Cannot divide by zero.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")

# Example usage
divide(10, 2)   # Output: Result: 5.0
divide(10, 0)   # Output: Error: Cannot divide by zero.
```

```python
#18. How would you implement a basic logging setup that logs to a file with rotation after 1MB?
#Code
import logging
from logging.handlers import RotatingFileHandler

# Set up a basic logging configuration with rotating file handler
def setup_logger():
    # Create a logger
    logger = logging.getLogger('my_logger')
    logger.setLevel(logging.DEBUG)  # Set the level to DEBUG to capture all log levels

    # Create a rotating file handler that logs to 'app.log' and rotates after 1MB (1048576 bytes)
    handler = RotatingFileHandler('app.log', maxBytes=1048576, backupCount=3)  # Keep 3 backup logs
    handler.setLevel(logging.DEBUG)  # Log everything at DEBUG level or higher

    # Create a formatter to specify the log format
    formatter = logging.Formatter('%(asctime)s - %(levelname)s - %(message)s')
    handler.setFormatter(formatter)

    # Add the handler to the logger
    logger.addHandler(handler)

    return logger

# Example usage
logger = setup_logger()

# Generate some log messages to test the rotation
for i in range(100000):
    logger.debug(f"This is a debug message number {i}")
    logger.info(f"This is an info message number {i}")
    logger.warning(f"This is a warning message number {i}")
    logger.error(f"This is an error message number {i}")
    logger.critical(f"This is a critical message number {i}")
```

```python
#20. How would you open a file and read its contents using a context manager in Python?
#Code
def read_file_contents(filename):
    try:
        # Using a context manager to open and read the file
        with open(filename, 'r') as file:  # Open the file in read mode
            content = file.read()  # Read the entire content of the file
            print(content)  # Print the content of the file
    except FileNotFoundError:
        print(f"Error: The file '{filename}' does not exist.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")

# Example usage
read_file_contents('example.txt')  # Replace with the name of your file
```

```python
#21. Write a Python program that reads a file and prints the number of occurrences of a specific word.
#Code
def count_word_occurrences(filename, word_to_find):
    try:
        with open(filename, 'r') as file:
            content = file.read()  # Read the entire file content
            
            # Convert the content to lowercase and split into words for case-insensitive matching
            words = content.lower().split()
            
            # Count the occurrences of the specific word
            word_count = words.count(word_to_find.lower())
            
            print(f"The word '{word_to_find}' occurs {word_count} times in the file.")
    
    except FileNotFoundError:
        print(f"Error: The file '{filename}' does not exist.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")

# Example usage
count_word_occurrences('example.txt', 'python')  # Replace with your file and word
```

```python
#22. How can you check if a file is empty before attempting to read its contents?
#Code
import os

def is_file_empty(filename):
    if os.path.getsize(filename) == 0:
        print(f"The file '{filename}' is empty.")
        return True
    else:
        print(f"The file '{filename}' is not empty.")
        return False

def read_file_if_not_empty(filename):
    if not is_file_empty(filename):
        with open(filename, 'r') as file:
            content = file.read()  # Read the entire content
            print("File contents:")
            print(content)

# Example usage
read_file_if_not_empty('example.txt')  # Replace with your file name
```
